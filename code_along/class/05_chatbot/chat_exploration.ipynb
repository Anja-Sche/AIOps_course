# Chatting with PydanticAI

```python
from pydantic_ai import Agent 
from model import nvidia_model

chat_agent = Agent(nvidia_model(),
                    instructions=("Be a joking programming nerd, always answer with a programming joke."
                    "Also add in some emojis to make it funnier."))

chat_agent
```

```python
result = await chat_agent.run("How is your day?")
result.output
```

```python
# NO MEMORY
result2 = await chat_agent.run("Whet did I just ask you?")
print(result2.output)
```

```python
result2.all_messages()
```

## Give a memory

```python
result = await chat_agent.run("Varför är tåget alltid sent i stockholm?")
print(result.output)
```

```python
result = await chat_agent.run("Gissa på vad vädret är här och säg vart här är?", message_history = result.all_messages())
print(result.output)
```

```python
result.all_messages()
```

```python
len(result.all_messages())
```

## Chat with our API endpoint

```python
import httpx

result = httpx.post(
    "http://127.0.0.1:8000/chat", 
    json={"question": "berätta en historia om en grå kanin som heter bella"}, 
    timeout=60)

result
```

```python
result.json().keys()
```

```python
message_history = result.json().get("message_history")
message_history[0]
```

```python
result = httpx.post(
    "http://127.0.0.1:8000/chat", 
    json={"question": "I föregående historia, vad hette kaninen?", "message_history": message_history}, 
    timeout=60)

result
```

# Structured output from text

```python
from model import nvidia_model
from pydantic_ai import Agent
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str 
    orginasation: str = Field("company or organisation")
    subject: str = Field("the subject the person has")

class Resiliance(BaseModel):
    title: str = Field("pick suitable headline from the input, don't hallucinate")
    description: str = Field("3-5 sentences summary of the content")
    contributors: list[Person]
    contact: str = Field("contact person for this event")

with open("raw_text.txt", encoding="utf-8") as file:
    raw_text = file.read()


agent = Agent(model=nvidia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions=("You are expert in summarizing swedish content in swedish.",
    "Take the provided content and summarize it according to output type"))

result = await agent.run(raw_text, output_type=Resiliance)
```

```python
contents = result.output
```

```python
class Course(BaseModel):
    course: str = Field("name of course")
    cours_description: str = Field("1-2 scentences of description of the course")
    points: int = Field("cours points, amount of YH-poäng")

class Program(BaseModel):
    title: str = Field("pick suitable headline from the input, don't hallucinate")
    description: str = Field("3-5 sentences summary of the program")
    competence: str = Field("2-3 scentences of the competence you get from the program")
    courses: list[Course]
    contact: str = Field("contact person for this program")


with open("raw_program.txt", encoding="utf-8") as file:
    raw_program = file.read()

with open("raw_program2.txt", encoding="utf-8") as file:
    raw_program2 = file.read()

agent = Agent(model=nvidia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions=("You are expert in summarizing swedish content in swedish.",
    "Take the provided content and summarize it according to output type"))

result = await agent.run(raw_program, output_type=Program)
```

```python
program_contents = result.output

program_contents
```

```python
result2 = await agent.run(raw_program2, output_type=Program)
program2_contents = result2.output

program2_contents
```

# Intro to llms using PydanticAI and Nvidia Build API

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
from dotenv import load_dotenv
import os

load_dotenv()


model = OpenAIChatModel(
    "nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API"],
    ),
)

agent = Agent(model, system_prompt=("You are a joking programming bot that will always ask questions back in a nerdy joke style",
    "Awnser with 3-5 sentences")
)

result = await agent.run("why is it foggy in göteborg")

result
```

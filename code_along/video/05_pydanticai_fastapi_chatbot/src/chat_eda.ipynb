# Explloring chatting with pydanticAI and FastAPI

```python
from pydantic_ai import Agent
from constants import MODEL
from dotenv import load_dotenv

load_dotenv()

chat_agent = Agent(
    MODEL,
    system_prompt="Be a joking programming nerd, always answer with a programming joke. Also add in some emojis to make it funnier.")

chat_agent
```

```python
result = await chat_agent.run("hello")
result
```

```python
print(result.output)
```

```python
result = await chat_agent.run("What did I just ask you?")
print(result.output)
```

```python
result.all_messages()
```

```python
result.all_messages()[1]
```

## Implementing memory

```python
result = await chat_agent.run("What is the weather in Stockholm")
result
```

```python
result = await chat_agent.run("What did I just ask you", message_history=result.all_messages())

print(result.output)
```

```python
len(result.all_messages())
```

## Try out endpoint

```python
import httpx

result = httpx.post(
    "http://localhost:8000/chat",
    json = {"question": "tell me a joke about a rabbit"}
)

result.json()["response"]
```

```python
msg_history = result.json()["message_history"]
msg_history
```

```python
result = httpx.post(
    "http://localhost:8000/chat",
    json = {"question": "what did I just ask you?", "message_history":msg_history}
)

result.json()["response"]
```
